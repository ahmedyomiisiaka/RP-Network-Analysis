## Folder Structure

/content/rp_network/
│
├── data/
│ ├── raw/ ← original downloads, never modified

**Note:** Large files (GraphML networks, raw TSV downloads) are NOT  
uploaded to GitHub. Only notebooks and scripts are version-controlled.

```python
# ─────────────────────────────────────────────────────────────────
# Notebook 00 — Cell 1: Create project folder structure
#
# This cell creates all folders needed for the entire project.
# Run this once at the start of every Colab session to ensure
# the folder structure exists before any notebook writes to it.
# ─────────────────────────────────────────────────────────────────

import os
from pathlib import Path
from datetime import datetime

# Define all directories
DIRS = [
    "/content/rp_network/data/raw",
    "/content/rp_network/data/processed",
    "/content/rp_network/data/results",
    "/content/rp_network/notebooks",
    "/content/rp_network/docs",
    "/content/rp_network/figures",
]

# Create them
for d in DIRS:
    Path(d).mkdir(parents=True, exist_ok=True)

print("✓ Project folder structure created:")
for d in DIRS:
    print(f"   {d}")
```

```python
# ─────────────────────────────────────────────────────────────────
# Notebook 00 — Cell 2: Verify Python environment
#
# Checks which libraries are already available in Colab and which
# will need to be installed in later notebooks.
# This is documentation — it does not install anything yet.
# ─────────────────────────────────────────────────────────────────

import sys

libraries_to_check = [
    ("requests",    "API calls to STRING, UniProt, Reactome"),
    ("pandas",      "Data tables and TSV files"),
    ("networkx",    "Network construction and analysis"),
    ("matplotlib",  "Plotting and figures"),
    ("seaborn",     "Statistical plots"),
    ("numpy",       "Numerical operations"),
]

libraries_to_install_later = [
    ("leidenalg",   "Leiden community detection — Stage 7"),
    ("python-igraph","Required by leidenalg — Stage 7"),
    ("bioservices", "UniProt and pathway API access — Stage 3"),
]

print(f"Python version: {sys.version.split()[0]}")
print()
print("Libraries available in this environment:")
print("-" * 50)

for lib, purpose in libraries_to_check:
    try:
        __import__(lib)
        print(f"  ✓  {lib:<15} — {purpose}")
    except ImportError:
        print(f"  ✗  {lib:<15} — NOT found (will install in relevant notebook)")

print()
print("Libraries to be installed in later notebooks:")
print("-" * 50)
for lib, purpose in libraries_to_install_later:
    print(f"  →  {lib:<15} — {purpose}")
```
